```python
import time 
import logging
import json 
import os
from datetime import datetime
from pathlib import Path
```

```python
os.getcwd()
```

```python
time_steps = datetime.now().strftime("%d_%m_%Y_%H_%M")
time_steps
```

```python
log_dir = "logs"

model_name = "GMF"

run_dir = Path(log_dir) / model_name / f"run_{time_steps}"
run_dir
```

```python
run_dir.mkdir(parents=True, exist_ok=True)
```

```python
config = {"batch_size" : 4, "out" : False}


config
```

```python
config_path = run_dir / "config.json"

with open(config_path, "w") as f:
    json.dump(config, f, indent=4)
    
log_type = "traning"  # "evaluation"
    
log_path = run_dir / f"{log_type}.log"

logger = logging.getLogger(f"{model_name}_{log_type}_{time_steps}")
logger
```

```python
logger.setLevel(logging.INFO)

formatter = logging.Formatter("%(asctime)s - %(name)s -%(levelname)s - %(message)s")
formatter
```

```python
fh = logging.FileHandler(log_path)
fh
```

```python
fh.setLevel(logging.INFO)
```

```python
ch = logging.StreamHandler()
ch.setLevel(logging.INFO)

ch
```

```python
formatter = logging.Formatter("%(asctime)s - %(name)s - %(levelname)s")
fh.setFormatter(formatter)
ch.setFormatter(formatter)

fh, ch
```

```python
logger.handlers
```

```python
logger.addHandler(fh)
logger.addHandler(ch)

logger
```
