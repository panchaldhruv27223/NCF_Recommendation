```python
import re
```

```python
def get_avg_time(log_file = ""):
    if log_file == "":
        print("Pass valid file")
        return 

    times = []


    pattern = re.compile(r"epoch no \d+: ([\d.]+)")

    with open(log_file, "r") as f:
        for line in f:
            match = pattern.search(line)
            if match:
                time_taken = match.group(1)
                times.append(float(time_taken))

    no_times = len(times)
    avg_time_each_epoch = sum(times)/no_times

    print(f"Avg time taken for each epoch is: {avg_time_each_epoch}")
```
